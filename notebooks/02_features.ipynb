# 02 · Feature engineering
Feature logic lives in `src/features.py`; edit it there, then `python -m src.pipeline --refresh features`. This notebook inspects `data/processed/features.parquet`.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src import config
from src.data_io import read_features
import json
```

```python
feats = read_features()
stats = json.loads(config.features_path().with_suffix(".stats.json").read_text())
print(stats)
feats.head()
```

## Candidate features before imputation

```python
pre = ["commission", "log_lamports", "mean_efficiency_pct", "median_efficiency_pct_raw",
       "efficiency_cv_raw", "performance_trend_raw", "num_epochs_recorded",
       "num_authorized_voters", "num_prior_voters"]
display(feats[pre].describe().T)
print(feats[pre].isna().sum())
```

```python
feats[pre].hist(bins=40, figsize=(14, 10)); plt.tight_layout(); plt.show()
```

```python
fig, axes = plt.subplots(len(pre), 1, figsize=(8, 1.4 * len(pre)))
for ax, f in zip(axes, pre):
    sns.boxplot(x=feats[f], ax=ax); ax.set_xlabel(""); ax.set_title(f, fontsize=9)
plt.tight_layout(); plt.show()
```

## Final model features (after imputation)

```python
display(feats[config.FEATURES].describe().T)
plt.figure(figsize=(8, 6))
sns.heatmap(feats[config.FEATURES].corr(), annot=True, cmap="coolwarm", center=0, fmt=".2f")
plt.title("Feature correlation"); plt.show()
```
